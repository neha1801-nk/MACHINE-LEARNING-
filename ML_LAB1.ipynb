SCENARIO 1: E-commerce Sales Data
Analyze a sample e-commerce dataset to understand product categories, sales trends, and missing information.
Dataset (Kaggle – Public):
https://www.kaggle.com/datasets/carrie1/ecommerce-data

IN-LAB TASKS:
•	Import necessary Python libraries.
•	Load the dataset into a Pandas DataFrame.
•	Inspect dataset using head(), tail(), info(), and describe().
•	Check for missing values.
•	Visualize sales per product using bar and line charts.



```python
#LAB 1 TASK 1
# NEHA.N.K | 24BAD078
import pandas as pd
import matplotlib.pyplot as plt
print("NEHA.N.K| 24BAD078")
# Loading dataset
print("Loading E-commerce dataset...")
data_set = pd.read_csv("data.csv", encoding="ISO-8859-1")


# Shows first 5 rows
print("\nFirst 5 rows of dataset:")
print(data_set.head())

# Shows last 5 rows
print("\nLast 5 rows of dataset:")
print(data_set.tail())

# Shows dataset info
print("\nDataset Info:")
print(data_set.info())

# Shows statistical summary
print("\nStatistical Summary:")
print(data_set.describe())

# Checking missing values
print("\nMissing Values per Column:")
print(data_set.isnull().sum())

# Group sales by product description and sum quantities
print("\nTop 10 Products by Sales:")
sales_per_product = data_set.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10)
print(sales_per_product)

# Bar chart
sales_per_product.plot(kind='bar')
plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.show()

# Line chart
sales_per_product.plot(kind='line')
plt.title("Sales Trend of Top Products")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.show()
```

SCENARIO 2: Hospital Patient Records
Examine a patient dataset to identify missing health metrics and visualize distributions.
Dataset (Kaggle – Public):
https://www.kaggle.com/datasets/uciml/pima-indians-diabetes-database

IN-LAB TASKS:
•	Load the dataset into Pandas.
•	Explore data structure and missing values.
•	Visualize glucose levels and age distribution using histograms and boxplots.
•	Identify patterns in health metrics


```python
import pandas as pd;
import matplotlib.pyplot as plt;

# NEHA.N.K | 24BAD078
# loading dataset
data_set=pd.read_csv("diabetes.csv",encoding="ISO-8859-1");
print("NEHA.N.K | 24BAD078");
# Shows first 5 rows
print("\nFirst 5 rows of dataset:")
print(data_set.head())

# Shows last 5 rows
print("\nLast 5 rows of dataset:")
print(data_set.tail())

# Shows dataset info
print("\nDataset Info:")
print(data_set.info())

# Shows statistical summary
print("\nStatistical Summary:")
print(data_set.describe())

# Checking missing values
print("\nMissing Values per Column:")
print(data_set.isnull().sum())

print("\nDisplaying Histogram of Glucose Levels...")
plt.hist(data_set["Glucose"], bins=20, color="skyblue", edgecolor="black")
plt.title("Glucose Level Distribution")
plt.xlabel("Glucose")
plt.ylabel("Frequency")
plt.show()

print("\nDisplaying Boxplot of Age Distribution...")
plt.boxplot(data_set["Age"])
plt.title("Age Distribution")
plt.ylabel("Age")
plt.show()

# Identifying hidden patterns through glucose and BMI
print("\nAverage Glucose by Diabetes Outcome (0 = No Diabetes, 1 = Diabetes):")
avg_glucose = data_set.groupby("Outcome")["Glucose"].mean()
print(avg_glucose)   # Prints average glucose for patients with and without diabetes

print("\nAverage BMI by Diabetes Outcome (0 = No Diabetes, 1 = Diabetes):")
avg_bmi = data_set.groupby("Outcome")["BMI"].mean()
print(avg_bmi)




```

SCENARIO 3: Housing Dataset
Investigate a housing dataset for missing features and relationships between house attributes.
Dataset (Kaggle – Public):
https://www.kaggle.com/datasets/harishkumardatalab/housing-price-prediction

IN-LAB TASKS:
•	Load the dataset and inspect columns.
•	Detect and summarize missing values.
•	Visualize correlations between numerical features using scatter plots and heatmaps.
•	Interpret trends for price prediction


```python
# NEHA.N.K | 24BAD078
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("NEHA.N.K | 24BAD078")

#Loading dataset
print("\nLoading Housing dataset...")
data_set = pd.read_csv("Housing.csv",encoding="ISO-8859-1");

print("\nColumn Names:")
print(data_set.columns)   # Print all column names

print("\nFirst 5 rows of dataset:")
print(data_set.head())

print("\nDataset Info:")
print(data_set.info())

print("\nStatistical Summary:")
print(data_set.describe())

print("\nMissing Values per Column:")
print(data_set.isnull().sum())   # Count missing values per column

# 4. Visualize correlations

# Scatter plot: Area vs Price
print("\nDisplaying Scatter Plot: Area vs Price...")
plt.scatter(data_set["area"], data_set["price"], alpha=0.5, color="blue", edgecolors="black")
plt.title("Area vs Price")
plt.xlabel("Area")
plt.ylabel("Price")
plt.show()

# Scatter plot: Bedrooms vs Price
print("\nDisplaying Scatter Plot: Bedrooms vs Price...")
plt.scatter(data_set["bedrooms"], data_set["price"], alpha=0.5, color="green", edgecolors="black")
plt.title("Bedrooms vs Price")
plt.xlabel("Bedrooms")
plt.ylabel("Price")
plt.show()

# Heatmap of correlations
print("\nDisplaying Correlation Heatmap...")
plt.figure(figsize=(10,6))
numeric_data = data_set.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_data.corr(), annot=True, cmap="coolwarm")
plt.title("Correlation Heatmap of Housing Features")
plt.show()

print("\nInterpreting Trends for Price Prediction:")
print("1. Area shows strong positive correlation with Price → larger houses cost more.")
print("2. Bedrooms and bathrooms also correlate with Price, but weaker than area.")
print("3. Heatmap helps identify which features are most useful for predicting price.")
```


SCENARIO 4: Banking Customer Data
Analyze banking data to detect missing account information and understand customer demographics.
Dataset (Kaggle – Public):
https://www.kaggle.com/datasets/ajay1735/customer-personality-analysis
IN-LAB TASKS:
•	Import dataset into Pandas.
•	Examine structure, null values, and column types.
•	Visualize age, income, and spending patterns using bar and box plots.
•	Summarize insights from visualizations.


```python
# NEHA.N.K | 24BAD078
print("NEHA.N.K | 24BAD078")

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/marketing_campaign.csv", sep="\t")

print("Dataset Info:\n")
print(df.info())
print("\nMissing Values:\n")
print(df.isnull().sum())

current_year = 2026
df["Age"] = current_year - df["Year_Birth"]
spend_cols = [

    "MntWines", "MntFruits", "MntMeatProducts",

    "MntFishProducts", "MntSweetProducts", "MntGoldProds"

]
df["Total_Spending"] = df[spend_cols].sum(axis=1)

# (a) Bar Plot – Age Distribution

age_bins = pd.cut(df["Age"], bins=10)
age_counts = age_bins.value_counts().sort_index()
plt.figure()
age_counts.plot(kind="bar")
plt.title("Age Distribution of Banking Customers")
plt.xlabel("Age Groups")
plt.ylabel("Number of Customers")
plt.show()

# (b) Box Plot – Income Distribution

plt.figure()
plt.boxplot(df["Income"].dropna())
plt.title("Income Distribution")
plt.ylabel("Income")
plt.show()

# (c) Box Plot – Total Spending Distribution

plt.figure()
plt.boxplot(df["Total_Spending"])
plt.title("Customer Spending Distribution")
plt.ylabel("Total Spending")
plt.show()
```
